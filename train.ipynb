```python
import torch
import torch.nn as nn
from bert.encoder import encoder_block
from bert.MLMhead import encoder_stack,Transformer
from bert.MLM import MLM
from tokenizers import Tokenizer
from pathlib import Path

try:
    
    BASE_DIR = Path(__file__).resolve().parent
except NameError:
    
    BASE_DIR = Path.cwd()

TOKENIZER_PATH = BASE_DIR / "Data_pipeline" / "bert_tokenizer.json"



tokenizer= Tokenizer.from_file(str(TOKENIZER_PATH))

vocab_size= tokenizer.get_vocab_size()

encoder_layers= [encoder_block(256,4,1024,eps=1e-5) for _ in range(4)]

enc_stack= encoder_stack(encoder_layers)

model= Transformer(enc_stack,256,vocab_size)

device= torch.device("cuda" if torch.cuda.is_available() else "cpu")

model= model.to(device)
```
